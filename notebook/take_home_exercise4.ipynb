## 2.Build a modeling dataset

```python
# Rename columns in standings tables to avoid name conflicts
driver_standings_renamed = driver_standings.rename(columns={
    "points": "driver_points",
    "position": "driver_standings_pos",
    "wins": "driver_wins"
})

constructor_standings_renamed = constructor_standings.rename(columns={
    "points": "constructor_points",
    "position": "constructor_standings_pos",
    "wins": "constructor_wins"
})

# Merge race-level information into results
model_df = results.merge(
    races[["raceId", "year", "round", "name", "date"]],
    on="raceId",
    how="left"
)

# Merge driver standings into the modeling dataset
model_df = model_df.merge(
    driver_standings_renamed[["raceId", "driverId", "driver_points", "driver_standings_pos", "driver_wins"]],
    on=["raceId", "driverId"],
    how="left"
)

# Merge constructor standings into the modeling dataset
model_df = model_df.merge(
    constructor_standings_renamed[["raceId", "constructorId", "constructor_points", "constructor_standings_pos", "constructor_wins"]],
    on=["raceId", "constructorId"],
    how="left"
)

# Rename qualifying position column
qualifying_renamed = qualifying.rename(columns={
    "position": "quali_position"
})

# Merge qualifying information into the modeling dataset
model_df = model_df.merge(
    qualifying_renamed[["raceId", "driverId", "constructorId", "quali_position"]],
    on=["raceId", "driverId", "constructorId"],
    how="left"
)
```

```python
# Create a binary target variable for podium finish
model_df["target"] = (model_df["positionOrder"] <= 3).astype(int)

# Check the shape and preview the merged dataset
print("Modeling dataset shape:", model_df.shape)
model_df.head()

# Inspect key modeling columns
selected_cols = [
    "raceId", "driverId", "constructorId", "year", "round",
    "grid", "positionOrder", "target",
    "driver_points", "driver_standings_pos", "driver_wins",
    "constructor_points", "constructor_standings_pos", "constructor_wins",
    "quali_position"
]

model_df[selected_cols].head()

# Check missing values in key columns
missing_summary = model_df[selected_cols].isna().sum().sort_values(ascending=False)
print(missing_summary)
```

```python
# Select feature columns for modeling
feature_cols = [
    "grid",
    "driver_points",
    "driver_standings_pos",
    "driver_wins",
    "constructor_points",
    "constructor_standings_pos",
    "constructor_wins",
    "quali_position",
    "year",
    "round"
]

# Keep only the columns needed for modeling
model_data = model_df[[
    "raceId",
    "driverId",
    "constructorId",
    "target"
] + feature_cols].copy()

# Fill missing values in numeric features
model_data["driver_points"] = model_data["driver_points"].fillna(0)
model_data["driver_wins"] = model_data["driver_wins"].fillna(0)
model_data["constructor_points"] = model_data["constructor_points"].fillna(0)
model_data["constructor_wins"] = model_data["constructor_wins"].fillna(0)

model_data["driver_standings_pos"] = model_data["driver_standings_pos"].fillna(99)
model_data["constructor_standings_pos"] = model_data["constructor_standings_pos"].fillna(99)
model_data["quali_position"] = model_data["quali_position"].fillna(99)
```

```python
# Define feature and target
X = model_data[feature_cols]
y = model_data["target"]

from sklearn.model_selection import train_test_split

# split data into training and testing sets
X_train, X_test, y_train, y_test, meta_train, meta_test = train_test_split(
    X,
    y,
    model_data[["raceId", "driverId", "constructorId"]],
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
# Print train and test shapes
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```
